**BUBBLE SORT**


```python
import time
numbers = []

# Take 5 inputs from the user
print("Enter 5 numbers:")
for i in range(5):
    num = int(input(f"Enter element {i + 1}: "))
    numbers.append(num)

# Start measuring execution time
start_time = time.perf_counter()

# Bubble Sort
n = len(numbers)

for i in range(n - 1):
    for j in range(n - i - 1):
        if numbers[j] > numbers[j + 1]:
            # Swap the elements
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

# Stop measuring execution time
end_time = time.perf_counter()

# Display the sorted list
print("\nSorted List:", numbers)

# Display execution time
execution_time = end_time - start_time
print(f"Execution Time: {execution_time:.10f} seconds")

# Display time complexity
print("\nTime Complexity:")
print("Best Case   : O(n)   (when the list is already sorted, with optimization)")
print("Average Case: O(n²)")
print("Worst Case  : O(n²)")
print("Space Complexity: O(1)")
```

**SELECTION SORT**

```python
import time

# Create an empty list
numbers = []

# Take 5 inputs from the user
print("Enter 5 numbers:")
for i in range(5):
    num = int(input(f"Enter element {i + 1}: "))
    numbers.append(num)

# Start measuring execution time
start_time = time.perf_counter()

# Selection Sort
n = len(numbers)

for i in range(n):
    min_index = i

    for j in range(i + 1, n):
        if numbers[j] < numbers[min_index]:
            min_index = j

    # Swap the minimum element with the current element
    numbers[i], numbers[min_index] = numbers[min_index], numbers[i]

# Stop measuring execution time
end_time = time.perf_counter()

# Display the sorted list
print("\nSorted List:", numbers)

# Display execution time
execution_time = end_time - start_time
print(f"Execution Time: {execution_time:.10f} seconds")

# Display time complexity
print("\nTime Complexity:")
print("Best Case   : O(n²)")
print("Average Case: O(n²)")
print("Worst Case  : O(n²)")
print("Space Complexity: O(1)")
```

**QUICK SORT**

```python
import time

# Function to perform Quick Sort
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[-1]  # Last element as pivot
    left = []
    right = []

    for i in arr[:-1]:
        if i <= pivot:
            left.append(i)
        else:
            right.append(i)

    return quick_sort(left) + [pivot] + quick_sort(right)

# Create an empty list
numbers = []

# Take 5 inputs from the user
print("Enter 5 numbers:")
for i in range(5):
    num = int(input(f"Enter element {i + 1}: "))
    numbers.append(num)

# Start measuring execution time
start_time = time.perf_counter()

# Perform Quick Sort
sorted_numbers = quick_sort(numbers)

# Stop measuring execution time
end_time = time.perf_counter()

# Display the sorted list
print("\nSorted List:", sorted_numbers)

# Display execution time
execution_time = end_time - start_time
print(f"Execution Time: {execution_time:.10f} seconds")

# Display time complexity
print("\nTime Complexity:")
print("Best Case   : O(n log n)")
print("Average Case: O(n log n)")
print("Worst Case  : O(n²)")
print("Space Complexity: O(log n) (recursive stack)")
```

**MERGE SORT**

```python
import time

# Function to perform Merge Sort
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        left = arr[:mid]
        right = arr[mid:]

        # Recursively sort both halves
        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        # Merge the two halves
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        # Copy remaining elements of left half
        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        # Copy remaining elements of right half
        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1

# Create an empty list
numbers = []

# Take 5 inputs from the user
print("Enter 5 numbers:")
for i in range(5):
    num = int(input(f"Enter element {i + 1}: "))
    numbers.append(num)

# Start measuring execution time
start_time = time.perf_counter()

# Perform Merge Sort
merge_sort(numbers)

# Stop measuring execution time
end_time = time.perf_counter()

# Display the sorted list
print("\nSorted List:", numbers)

# Display execution time
execution_time = end_time - start_time
print(f"Execution Time: {execution_time:.10f} seconds")

# Display time complexity
print("\nTime Complexity:")
print("Best Case   : O(n log n)")
print("Average Case: O(n log n)")
print("Worst Case  : O(n log n)")
print("Space Complexity: O(n)")
```

**INSERTION** **SORT**

```python
import time

# Insertion Sort Function
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key


# User Input
n = int(input("Enter the number of elements: "))
arr = []

print("Enter the elements:")
for i in range(n):
    arr.append(int(input()))

# Measure Execution Time
start_time = time.perf_counter()

insertion_sort(arr)

end_time = time.perf_counter()

# Output
print("\nSorted Array:", arr)

# Time Complexity
print("\nTime Complexity:")
print("Best Case   : O(n)")
print("Average Case: O(n^2)")
print("Worst Case  : O(n^2)")

# Execution Time
print(f"\nExecution Time: {(end_time - start_time):.10f} seconds")
```
